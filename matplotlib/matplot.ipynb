```python
#1 labeled graph 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
x = np.array([2004,2005,2006,2007,2008,2009])
y = np.array([1,2,3,4,5,6])
plt.xlabel("Year")
plt.ylabel("Age")
plt.plot(x,y)
```

```python
#2 draw a line with labels and title 
plt.xlabel("Year")
plt.ylabel("Age")
plt.title("Age According to year")
plt.grid()
plt.plot(x,y,marker="o")
```

```python
df = pd.read_csv("matplotlib/test.txt",sep="  ")
print(df)
plt.xlabel("X-Values")
plt.ylabel("Y-Values")
plt.plot(df["a"], df["b"])
```

```python
#3 and 4
df = pd.read_csv("matplotlib/alphabet.csv")
print(df)
print(df.columns)
plt.grid()
plt.plot(df["Open"],df["Close "],label = "open and close")
plt.plot(df["High"],df["Low"],marker = "o", label = "high and low")
plt.legend()
```

```python
#5, 6 ,7 different lines with different styles
plt.grid()
plt.plot(df["Open"],df["Close "],label = "open and close",color = "g", linewidth = "5", linestyle = "--")
plt.plot(df["High"],df["Low"],marker = "o", label = "high and low" , color = "hotpink", linewidth = "6", ms = "10",mec = "r", markerfacecolor = "g")
plt.legend()
```

```python
#9 change the axis values
e = np.array([1,2,3,4,5,6])
d = np.array([10,20,30,40,50,60])
plt.plot(e,d)
print(plt.axis())
# plt.xlim(0,10)
# plt.ylim(0,100)
plt.axis([0,10,0,100])
plt.show()
```

```python
#10 add markers at the specified points
plt.plot(e,d,marker = "*",mfc = "red",ms = "15",mec = "g")
```

```python
#11
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
c = np.array([9,10,11,12])
d = np.array([13,14,15,16])
#x axis and y axis and format string = color+marker+linestyle
plt.plot(a,d,"b*--",c,d,"ro-")
 
```

```python
#12 multiple types of charts
plt.plot(a,b)
plt.figure()
plt.scatter(c,d)
plt.figure()
plt.bar(a,b)
plt.figure()
plt.barh(a,b)
plt.figure()
plt.pie(a,labels=["apple","banana","fruit","papaya"])
plt.legend()
```

```python
#13 linestyle -, width .5. and color blue
plt.plot(df["Date"],df["Close "],color = "r")
plt.grid(ls = "--",color = "b",linewidth = "5")
```

```python
#14 grid customization
plt.plot(df["Date"],df["Close "],color = "r")
plt.xlabel("Date")
plt.ylabel("Close")
plt.minorticks_on()
plt.grid(True,which="major",lw = "1",ls = "--",color = "b")
plt.grid(True,which="minor",lw="0.75",ls = "-",color = "g")
```

```python
#15
plt.subplot(2,3,1)
plt.plot(df["Date"],df["Close "],color = "r")
plt.title("line plot")

plt.subplot(2,3,2)
plt.scatter(df["Date"],df["Close "],color = "r")
plt.title ("Scatter plot")

plt.subplot(2,3,3)
plt.bar(df["Date"],df["Close "],color = "r")
plt.title ("Bar plot")

plt.subplot(2,3,4)
plt.pie(df["Close "],labels = ["772.559998" ,"776.429993","776.469971","776.859985","775.080017" ])
plt.title ("pie plot")

plt.subplot(2,3,5)
plt.hist(df["Close "])
plt.title ("pie plot")

plt.subplots_adjust(hspace = 2, wspace = 3)
plt.show()



```
