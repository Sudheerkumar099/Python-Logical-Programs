```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
#1
prog_lang = np.array([ "Java", "Python", "PHP", "JavaScript", "C#", "C++"])
Popularity = np.array([ 22.2, 17.6, 8.8, 8, 7.7, 6.7 ])
plt.xlabel("Programing languages")
plt.ylabel("Popularity")
plt.bar(prog_lang,Popularity,color = ["r","g","yellow","b","orange","pink"])
```

```python
#2 horizontal representation
plt.ylabel("Programing languages")
plt.xlabel("Popularity")
plt.barh(prog_lang,Popularity,color = ["r","g","yellow","b","orange","pink"])
```

```python
#3 same color
plt.bar(prog_lang,Popularity,color = ["r"])
```

```python
#4 different colors
plt.bar(prog_lang,Popularity,color = ["r","g","yellow","b","orange","pink"])
```

```python
#5 and 6  display the popularity upside of the bar and border is blue
plt.bar(prog_lang,Popularity,color = ["r","g","yellow","b","orange","pink"], edgecolor = "b")
```

```python
#10
means_men = (22, 30, 35, 35, 26)
means_women = (25, 32, 30, 35, 29)
groups = np.arange(len(means_men))
bar_width = 0.35
plt.bar(groups, means_men, width=bar_width, label='Men')
plt.bar(groups + bar_width, means_women, width=bar_width, label='Women')
plt.legend()
```
