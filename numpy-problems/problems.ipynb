```python
#5 middle as zero
arr = np.ones((5,5))
arr[1:len(arr)-1:1, 1:len(arr)-1:1]=0
print(arr)

```

```python
#6 border as zero
arr = np.random.randint(1,10,(10,10))

arr[0,0:]=0
arr[len(arr)-1,0:]= 0
for i in range(1,len(arr)-1):
    arr[i,0]= 0
    arr[i,len(arr[1])-1] = 0
print(arr)


arr2 = np.random.randint(1,10,(5,5))
rows, cols = arr2.shape
vborder = np.zeros((1,cols))
arr2 = np.vstack((vborder, arr2 ,vborder))
hborder = np.zeros((rows+2,1))
arr2 = np.hstack((hborder,arr2,hborder))
print(arr2)

```

```python
#7 chess board
board = np.ones((8,8),dtype=int)
board[0::2,0::2] = 0
board[1::2,1::2] = 0
print(board)
```

```python
#9 concatinate elements
arr = np.array([10,20,30,40])
arr2 = np.array([50,60,70,80,90])
arr4 = np.append(arr,arr2)
print(arr4)

arr3 = np.concatenate([arr,arr2],axis=0)
print(arr3)
```

```python
# 10 real and imaginary part
arr= np.array([1.00000000+0.j, 0.70710678+0.70710678j])
real = arr.real
imag = arr.imag
print("Real part of the array:  \n",real)

print("Imaginary part of the array: \n",imag)

```

```python
#12 unique elements
arr1 = np.array( [0,10,20,40,60])
arr2 = np.array([10, 30, 40, 50, 70, 90])
arr3 = np.intersect1d(arr2,arr1)
print(arr3)

```
