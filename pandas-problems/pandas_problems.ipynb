```python
#1 create a 1d array and print
import pandas as pd
import numpy as np
Numbers = np.array([10,20,90])
s1 = pd.Series(Numbers)
print(s1)
```

```python
#2 convert the 1d array to a list and print
lst = s1.tolist()
print(lst)
print(type(lst))
```

```python
#3 add, subtract, multiple and divide two Pandas Series.
s1 = pd.Series([2, 4, 6, 8, 10])
s2 =  pd.Series([1, 3, 5, 7, 9])
# print(s1 + s2)
print(s1.add(s2))
# print(s1 - s2)
print(s1.sub(s2))
# print(s1 * s2)
print(s1.mul(s2))
# print(s1 / s2)
print(s1.div(s2))

```

```python
#4 First array elements raised to powers from second array
s1 = pd.Series([0,1,2,3,4,5,6])
s2 = pd.Series([1,2,3,4,5,6,7]) 
print(s1 ** s2)
print(s1.pow(s2))
x = s1.index
print(s1.apply((lambda x : s1[x] ** s2[x])))
```

```python
# Pandas (Intermediate)
# A DataFrame contains date, city, temperature, and humidity.
# Task:
# Filter data for a single city
# Compute rolling 3-day average temperature
# Detect days where humidity increased compared to the previous day
data = {
    "date": [
        "2025-01-01", "2025-01-01", "2025-01-01",
        "2025-01-02", "2025-01-02", "2025-01-02",
        "2025-01-03", "2025-01-03", "2025-01-03"
    ],
    "city": [
        "Bengaluru", "Hyderabad", "Chennai",
        "Bengaluru", "Hyderabad", "Chennai",
        "Bengaluru", "Hyderabad", "Chennai"
    ],
    "temperature": [24, 26, 28, 23, 27, 29, 22, 25, 30],
    "humidity": [68, 60, 72, 70, 58, 75, 65, 62, 78]
}

df = pd.DataFrame(data)
blr = df[df["city"] == "Bengaluru"]
hyd = df[df["city"] == "Hyderabad"]
chn = df[df["city"] == "Chennai"]
print("Filtered Data of Bengalulu")
print(blr)
print("Filtered Data of Hyderabad")
print(hyd)
print("Filtered Data of chennai")
print(chn)

# average temperature 
print(blr["temperature"].mean())
print(chn["temperature"].mean())
print(hyd["temperature"].mean())
```
