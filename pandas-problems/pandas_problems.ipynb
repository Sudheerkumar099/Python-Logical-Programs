```python
#1 create a 1d array and print
import pandas as pd
import numpy as np
Numbers = [10,20,90]
s1 = pd.Series(Numbers)
print(s1)
```

```python
#2 convert the 1d array to a list and print
lst = s1.tolist()
print(lst)
print(type(lst))
```

```python
#3 add, subtract, multiple and divide two Pandas Series.
s1 = pd.Series([2, 4, 6, 8, 10])
s2 =  pd.Series([1, 3, 5, 7, 9])
# print(s1 + s2)
print(s1.add(s2))
# print(s1 - s2)
print(s1.sub(s2))
# print(s1 * s2)
print(s1.mul(s2))
# print(s1 / s2)
print(s1.div(s2))

```

```python
#4 First array elements raised to powers from second array
s1 = pd.Series([0,1,2,3,4,5,6])
s2 = pd.Series([1,2,3,4,5,6,7]) 
# print(s1 ** s2)
print(s1.pow(s2))
```

```python
#5 create and display a DataFrame
exam_data = {'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 
'Matthew', 'Laura', 'Kevin', 'Jonas'], 
'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19], 
'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1], 
'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']} 
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j'] 
df = pd.DataFrame(exam_data,index=labels)
print(df)
```
