```python
import plotly.express as px
import numpy as np
import pandas as pd
px.line(x = [1,2,3,4] , y = [20,30,40,50])
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [120, 150, 170, 160, 180, 200]
px.line(x = months, y = sales)
```

```python
px.bar(x = months, y = sales)
```

```python
px.pie(values = sales, names = months,title = "Months and sales contribution")
```

```python
px.histogram(x = sales, nbins = 5 , title = "sales")
```
